## Strings II: case sensitivity and filepaths
I said above that strings are meaningless to the computer in *most* cases. There are two important notes to be made here: one, that the computer does not generally know that uppercase and lowercase letters have similar meanings; and second, that filepaths are an important exception.

### CaSe SeNsiTivIty
As a spoken or written language, English is generally case insensitive. To an English-speaking human, the following sentences all have basically the same meaning:

- The quick brown fox jumped over the lazy dog.
- THE QUICK BROWN FOX JUMPED OVER THE LAZY DOG.
- The Quick Brown Fox Jumped Over The Lazy Dog.
- The quick brown Fox jumped over the lazy Dog.
- The QuICK broWN FOx juMPeD oVEr ThE lAzy doG.

In Python, however, an uppercase or lowercase letter do not have the same meaning. As such, the five sentences above would be almost indistinguishable from one another! In fact, we can prove this with two simple True/False Boolean checks using the equality operator, `==`:

```
# The (==) operator returns either True or False.
# It essentially means, "check if the object on the left is equal to the object on the right."
# If the two objects are exactly the same, the result will be True.

"The quick brown fox jumped over the lazy dog." == "The quick brown fox jumped over the lazy dog."
"The quick brown fox jumped over the lazy dog." == 'The quick brown fox jumped over the lazy dog.'
"The quick brown fox jumped over the lazy dog." == "THE QUICK BROWN FOX JUMPED OVER THE LAZY DOG."
```

More complex indexing is accomplished by adding additional numbers separated by the `:` operator. Up to three numbers can be added in this way, following the  `[start:end:step]` format. For example, we can return the first through second items using `[0:2]`. This roughly translates to "start from index position zero, up to but not including index position 2."

Note that because Python starts counting from zero, index position zero is the first item, and index position two is the third item. I didn't make the rules.

````python
# this is a simple list of sample names
["OSF2021.22-iNat91369087-G11", "OSF2021.27-iNat91369529-A7", "OSF2021.41-iNat91373178-A1", "OSF2021.45-iNat91406034-G12"][0:2]
````

The third space in the index, the step, indicates how many positions to count by. With our example list, you could use `[0:3:2]` to say "start from index position zero, up to but not including index position 4, counting by twos."

Note that because Python starts counting at zero, index position 4 is actually the fifth position in the list. Since our list only has 4 items, this includes all of them.

````python
# this is a simple list of sample names
["OSF2021.22-iNat91369087-G11", "OSF2021.27-iNat91369529-A7", "OSF2021.41-iNat91373178-A1", "OSF2021.45-iNat91406034-G12"][0:3:2]
````

You don't actually need to use all the options in the `[start:end:step]` format; you simply include the `:` operator but omit the number. Skipping `start` will start at the beginning (position zero); skipping `end` will continue to the end of the list; and skipping `step` will count by ones.

## Strings III: f-strings, a fuper fpecial ftyle of ftring
Python also includes a very useful type of string, called the **f-string** (the "f" stands for "formatted"). String formatting lets you replace portions of the string with variables, setting the stage for dynamic things like file names, paths and directories, *et cetera.* To define a simple f-string, you preface it with the letter `f` (or `F`; this is one of the few case-insensitive things in Python) and enclose variables in curly braces (`{}`).

(click `>` at the upper left corner of this chunk to expand or collapse this section)

As a purely hypothetical example, let's imagine that we were working toward:
- Building a string including three instances of a sample name,
- That the first ended with a forward primer and file extension,
- That the second ended with a forward primer and the same file extension,
- And that the third was to receive no suffix at all.

Let's assume (again, purely hypothetically and for the sake of example) that our forward and reverse primers were called ITS1F and ITS4, and that we separated them from the sample name with a single dash. So, this portion of our string would look something like this for our first sample:

```
"OSF2021.22-iNat91369087-G11-ITS1F.ab1 OSF2021.22-iNat91369087-G11-ITS4.ab1 OSF2021.22-iNat91369087-G11"
```

Since the sample name appears three times in the string (and we know we have other samples that would receive the same treatment), it is a prime candidate for replacement with a variable. In this example, we'll use `sample_ID` as our variable name, build the f-string, and print it for posterity.

```
sample_ID = "OSF2021.22-iNat91369087-G11" # note that this must be a text string

print(f"{sample_ID}-ITS1F.ab1 {sample_ID}-ITS4.ab1 {sample_ID}")
```

````python
sample_ID = "OSF2021.22-iNat91369087-G11" # note that this must be a text string

print(f"{sample_ID}-ITS1F.ab1 {sample_ID}-ITS4.ab1 {sample_ID}")
````

## Introduction to Iterables: Baby's first `for` loop
In Python, certain types of object fall into a category called **iterables**. These can be logically broken into discrete, single units, upon which a command can act; each time this action is performed, it is called an **iteration.**

(click `>` at the upper left corner of this chunk to expand or collapse this section)

The `for` loop is one of the most useful ways to iterate over an iterable. It is formatted as follows: `for X in Y: Z`

Where...
- `for` is the statement initiating the loop
- `X` is a placeholder variable indicating each item in the iterable (if you want to look very clever, you can call this the **iterand**)
- `in` is an operator indicating that the iterand is in the iterable
- `Y` is the iterable itself
- `:` indicates that the iterable and iterand are defined, and the instructions for each iteration follow, and
- `Z` is the code to be repeated, on every iterand, for every iteration.

Note that the code (`Z`) is usually placed on its own line, indented with tabs or spaces. The amount and type of indentation is less important than consistency.

```
for X in Y:
  Z
```

Lists are one of the most common iterables.

```

  print(fruit)
```

````python
for fruit in ["Mango", "Rambutan", "Zucchini"]:
  print(fruit)
````

In the previous section, we mentioned that you can assign a variable to represent a list. Note that in most cases, the iterable and iterand should not be the same variable (`for fruit in fruits` is good practice; `for fruit in fruit` may work, but probably not as intended).

````python
fruits = ["Mango", "Rambutan", "Zucchini"]

for fruit in fruits:
  print(fruit)
````

You can also iterate over numeric ranges, which can be generated using the range() function:


````python
for number in range(0, 5):
  print(number)
  print(number + 1)
````

Strings are a type of iterable. You can also slice and index them like lists.

```

  print(letter)
```

In practice, you probably won't iterate over strings much (although things like *lists* of strings come up a lot). Slicing and indexing strings, however, can be quite handy.

## Bashing loudly: redirecting standard out and standard error
When Bash runs on its own (that is, without shell magic or flags like `!`), it doesn't assume that we need to know what it's doing; it just does it. We lab rats, however, like to know whether things worked or not. Fortunately, Bash does keep track of itself in in two places; these are called the **standard output** and **standard error**. In general, things that work go to standard output, while standard error gets errors and crash logs. This isn't always true, however, so it's common practice to redirect both to somewhere we can see them. The easiest way to visualize them is the `print()` function.

When `subprocess.run()` executes a command, it produces a data packet that can be saved as a variable. We can see this with `echo`, the Bash version of `print()`:


This is Python telling us: "I have told the OS to run the command, `echo "Hello, world!"`, and it did so successfully. This is all well and good, but the actual "Hello, world!" output was sent to the standard output, which vanishes into the aether.

To save this output, all we have to do is the following:
1. Add the `capture_output=True` and `text=True` arguments
2. Assign the data packet `subprocess` creates as a variable (technically optional, but makes your code much easier to read and write)
3. Use the `.stdout` and `.stderr` **attributes** on that variable (hold on)
4. Print those outputs

```
# don't forget to import your modules
import subprocess

# a simple echo (bash's version of print())
output = subprocess.run(["echo", "Hello, world!"], capture_output=True, text=True)

# printing the stdout and stderr associated with output
print(output.stdout)
print(output.stderr) # this may be blank, if there were no errors
```

````python
# don't forget to import your modules
import subprocess

# a simple echo (bash's version of print())
output = subprocess.run(["echo", "Hello, world!"], capture_output=True, text=True)

# printing the stdout and stderr associated with output
print(output.stdout)
print(output.stderr) # this may be blank, if there were no errors
````

Essentially, `capture_output=True` tells `subprocess` to look for any feedback that would normally be send to the standard output or standard error outputs.

`text=True` is optional; it tells `subprocess` to render things like line breaks and tabs (instead of printing their programming language equivalents, which typically look like "\n" or "\t"), which makes things a little easier to read. As always, try tinkering: remove it and see how it affects the output.

### Dot notation II: attributes
The notation we used above looks very similar to the `module.function()` notation we saw earlier. However, this is something new!

When `subprocess` runs, it creates an execution log based on the command being called. That log may have multiple parts, including the standard output and standard error. These parts are called **attributes**. In this case, dot notation (`.stdout` and `.stderr`) is used to target specific attributes of the log, often for `print()`-ing or saving to a file.

# 5) Why even bother with Python? Why not just use Bash for everything?

The biggest advantage of Python is that it was designed from the ground up to be human-readable and intuitive to skim. Bash is based on a much older language, which makes heavy use of symbols; it is very concise, but was developed when readability and intuitiveness were not major priorities.

The following code block is the equivalent of our Python code above, but written in Bash. The preview is heavily commented, but the code cell below is not (for an apples-to-apples comparison).

```
# bash cell magic; this entire cell will run Bash, instead of Python
%%bash

# our samples iterable, as a bash array (similar to a Python list)
# note that bash arrays don't use commas (no problem)
# also note that we can't use spaces to make things visually tidy (boo)
samples=("OSF2021.22-iNat91369087-G11" "OSF2021.27-iNat91369529-A7" "OSF2021.41-iNat91373178-A1")

# sample is the nickname we're giving our iterand
# the quotes "" allow the strings in the array to contain spaces
# the dollar sign $ indicates that the word "samples" is a variable
# the curly braces {} allow parameter expansion (doing things with items in our array)
# the square brackets [] indicate that the iterable is indexed (each item has a unique position)
# the at sign @ is a wildcard that means "anything and everything in the array"
# the semicolon ; indicates that we have defined our iterable and iterand
for sample in "${samples[@]}";
 # "do" this starts the command block. indentation is not required.
 do
   # we save our command as an array, similar to Python's subprocess.run
   # the dollar signs indicate our iterand variable
   # we won't use "command" as our variable, because that word can cause conflicts in Bash
   cmd=("/content/tracy-v0.9.1-linux-amd64" "consensus" "$sample-ITS1F.ab1" "$sample-ITS4.ab1" "-o" "$sample")
   
   # we run our command. no ! is required here, because we're in Bash
   # the symbols here indicate that the command is a variable, which is an array, etc as above
   "${cmd[@]}"

 done # this ends the command block
 ```
